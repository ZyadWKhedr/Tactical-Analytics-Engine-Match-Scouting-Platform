# 03 - Tactical Features, Zone Partitioning & Baselines

## Overview
This notebook extracts match-level tactical features, spatial zone metrics (6×5 grid), sequence tempos, entropy, and team/player season baselines with Empirical-Bayes shrinkage.


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Extracting spatial zones & tactical baselines...")
```

## 1. Spatial Zone Partitioning (6x5 Defensive Coordinates)


```python
def assign_zones(df):
    """Assigns actions to 6 bands x 5 channels."""
    df = df.copy()
    band_edges = np.linspace(0, 105, 7)
    channel_edges = np.linspace(0, 68, 6)
    
    df['band'] = np.clip(np.digitize(df['start_x'], band_edges) - 1, 0, 5)
    df['channel'] = np.clip(np.digitize(df['start_y'], channel_edges) - 1, 0, 4)
    df['zone_id'] = df['band'] * 5 + df['channel']
    return df

actions_path = os.path.join('data', 'actions.csv')
if os.path.exists(actions_path):
    df_act = pd.read_csv(actions_path, nrows=5000)
    df_zoned = assign_zones(df_act)
    print("Assigned zones summary:")
    print(df_zoned[['band', 'channel', 'zone_id']].head())
```
